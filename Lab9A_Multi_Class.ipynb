# Dataset cleaning and information

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Read Dry Bean csv file
df = pd.read_csv("Dry_Bean_Dataset_multiclass.csv")

# Displaying the dataset
display(df.head())
print()

# Displaying dataset description
print("Dataset description: \n")
display(df.describe())
print()

# Finding the missing values and dropping them
print("Shape before dropping missing values: ", df.shape)
print("Columns with missing values: ")
display(df.isnull().sum())
print()
df.dropna(inplace = True)
print("Columns after dropping missing values: ")
display(df.isnull().sum())
print("Shape after dropping missing values: ", df.shape)
print()

# Checking how many classes exist
print("Number of unique classes: ", sep = '')
display(df["Class"].unique())
print()

# Checking how many classes exist
print("Class distribution: ")
display(df["Class"].value_counts())
print()

# Since the machines don't understand strings, we will use the encoder to give each class an integer label. Sort of like a key in a dict
# Creating label encoder object
le = LabelEncoder()

# The fit_transform function is used to learn and assign numbers to the unique values of an array (in this cases its the output class)
df["Class"] = le.fit_transform(df["Class"])

# Checking the new class numbering
display(df["Class"].unique())
```

# Dataset splitting and training

```python
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier

# Assigning inputs and outputs
x = df.loc[:, df.columns != "Class"]
y = df.loc[:, df.columns == "Class"]

# Training and testing split using 80/20
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size = 0.2,
    train_size = 0.8,
    random_state = 2,
    )

# Creating MLP model and specifying its parameters
model = MLPClassifier(
    solver = "adam",
    alpha = 0.001,
    hidden_layer_sizes = (5),
    activation = "logistic",
    max_iter = 100,
    tol = 0.0001
    )

# Now we will train our model
model.fit(x_train, y_train)

# Now we will test the model
y_pred = model.predict(x_test)
```

# Results and metrics

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

# Confusion matrix
print("Confusion matrix: \n")
print(confusion_matrix(y_test, y_pred))
print()
print("Classification report: \n")
print(classification_report(y_test, y_pred))
print("Accuracy: ", accuracy_score(y_test, y_pred))
print("Precision: ", precision_score(y_test, y_pred, average = "weighted"))
print("Recall: ", recall_score(y_test, y_pred, average = "weighted"))
print("F1 Score: ", f1_score(y_test, y_pred, average = "weighted"))

```
